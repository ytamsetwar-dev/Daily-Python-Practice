# 2. String Data Type

### String Slicing Questions

#### 1) Print First 5 Number By User Input

```python
name = input("Enter String: ")

print("First 5 character: ",name[:5])
```

#### 2) Print Last 4 Number By User Input

```python
print("Last 4 Character: ",name[-4:])
```

#### 3) Print character from index 2 to 8 (excluding 8)

```python
print("char from index 2 to 8 : ",name[2:8])
```

#### 4) Print Every 2nd Character of a string

```python
print(" every 2nd char of string: ",name[::2])
```

#### 5) Print first name and last name of full name

```python
full_name = input("Enter Full Name: ")

first_name= full_name.find(" ")
last_name = full_name.rfind(" ")


print("First name:",full_name[:first_name])
print("Last name: ",full_name[last_name+1:])
```

#### 23. Take any string from user.
##### • Print the string in reverse order using slicing
##### • Convert it to title case
##### • Check if the reversed string is same as original (Palindrome check)

```python
user = input("Enter String: ")


rev_str = user[::-1]
print(rev_str)

print(user.upper())

print("Palindrome" * (user == rev_str) or "not palindrome")
```
