### Basic Idea To Solve the Task

#### 🟢 Basic List Comprehensions (1–10)

```python
# 1. Numbers from 1 to 10

[x for x in range(1, 11)]
```

```python
# 2. Squares from 1 to 10

[x**2 for x in range(1, 11)]
```

```python
# 3. Even numbers from 1 to 20

[x for x in range(1,30) if x % 2 == 0]
```

```python
# 4. Loop conversion

result = [i * 3 for i in range(10)]
result
```

```python
# 5. Output prediction
Output: [0, 1, 2, 3, 4]
```

```python
# 6. First 5 multiples of 7
[x * 7 for x in range(1, 6)]
```

```python
# 7. Strings to uppercase
[s.upper() for s in ["apple", "banana", "cherry"]]
```

```python
# 9. Numbers from 0 to 9

[x for x in range(10)]
```

```python
# 10. Only positive numbers
[x for x in [-2, -1, 0, 1, 2, 3] if x > 0]
```
