# Lambda Function

```python
# Lambda Fucntion is the anonymous function 
```

```python
# Suntax: 

lambda args : expression 

# Normal function

def function_name(args)
```

```python
# lambda function to find square

square = lambda x: x * x
square(5)
```

```python
# lambda function to find cube

cube = lambda x: pow(x,3)
cube(5)
```

```python
# no args

A = lambda : "Hello"
A()
```

```python
# one args

B=lambda x: x * x
B(5)
```

```python
# two args

B=lambda x,y : x + y
B(2,3)
```

```python
# You can use as much as aruguments....
```

```python
# Default Args

C = lambda x=5 : x*x
print(C())            # print by default arg
print(C(10))          # print by user arg
```

```python
# You can take 1 by user and 1 by by default also..
```

```python
# Multi Arg

multi=lambda *args : sum (args)
multi(1,2,3,4,5)
```

```python
# Multi args and multi key args

f=lambda *args, **kwargs : (args,kwargs)
f(1,2,3,4,a=20,b=30)
```

```python
# imp if else

# syntax: 
"Exp" if condition else 
```

```python
# Ex:
voting = lambda x: "Eligible" if x >= 18 else "Not Eligible"
voting(50)
```

```python
# Nested if else statement

A = lambda x: "A" if x >= 90 else "B" if x >= 60 else "Fail"
A(65)
```

```python
# lambda using list
B = lambda x: [x+x]
B(10)
```

```python
# lambda using set
B = lambda x: {x+x}
B(10)
```

```python
# lambda using dictionary

dicts = lambda name,age : {"Name":name,"Age":age}

dicts("Yash",23)
```

```python
# Lambda inside map function (imp)

num = [1,2,3,4,5,6]
result = list(map(lambda x:"Even"if x % 2 == 0 else "odd",num))
result
```

```python
# Lambda inside filter function (imp)

num = [1,2,3,4,5,6]
result = list(filter(lambda x:  x % 2 == 0,num))
result
```
