```python
class Bank_management:

    # create constructor
    def __init__(self,acc_no,balance,branch):

        # create the attributes and store
        self.Account_number=acc_no
        self.Balance=balance
        self.Branch=branch

    # display all the attributes

    # method 1
    def display_bank_detail(self):
        print(f"Account Number:{self.Account_number}")
        print(f"Balance:{self.Balance}")
        print(f"Branch:{self.Branch}")

    # method 2
    def deposit_money(self,amt):
        self.Balance += amt
        print(f"Deposit {amt} Successfully Done")

    # method 3
    def withdraw_money(self,amt):
        self.Balance -= amt
        print(f"Withdraw money {amt}")

# create the class object
bank=Bank_management(12345,20000,"Ameerpet")

# call the deposit method
bank.deposit_money(10000)

# call the withdraw method
bank.withdraw_money(20000)

# call display method
bank.display_bank_detail()
```
