### 9. Simple Calculator

```python
a = int(input("Enter Number 1:"))
b = int(input("Enter Number 2:"))

operator = (input("Enter Operation: "))

if (operator == "+"):
    result= a+b
    print(result)

if (operator == "-"):
    result= a-b
    print(result)
    
if (operator == "*"):
    result= a*b
    print(result)

if (operator == "/"):
    result= a/b
    print(result)

if (operator == "//"):
    result= a//b
    print(result)

if (operator == "%"):
    result= a%b
    print(result)

if (operator == "**"):
    result= a**b
    print(result)
    



```

### 10. Temperature Converter

```python
choice = input("Enter Choice: ")

if(choice=="Celcious to Fahrenheit"):
    C=int(input("Enter Celcious: "))
    F=(C*9/5)+32
    print(F)
if(choice == "Fahrenheit to Celcious"):
    f=int(input("Enter Fahrenheit: :"))
    celsius = (f - 32) * 5 / 9
    print(celsius)
    
     
     
                 

```

### 12. Student Grade Calculator

```python
sub1=int(input("Enter subject 1 marks: "))
sub2=int(input("Enter subject 2 marks: "))
sub3=int(input("Enter subject 3 marks: "))
sub4=int(input("Enter subject 4 marks: "))
sub5=int(input("Enter subject 5 marks: "))

percentage=(sub1+sub2+sub3+sub4+sub5)/5
print(percentage)

if(percentage >=90):
    grade = "A+"
    print(grade)

elif(percentage >=80):
    grade = "A"
    print(grade)
    
elif(percentage >=70):
    grade = "B"
    print(grade)
    
elif(percentage >=60):
    grade = "C"
    print(grade)
    
else:
    print("Fail")



```
