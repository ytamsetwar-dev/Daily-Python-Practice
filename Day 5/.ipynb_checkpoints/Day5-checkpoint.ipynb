## Easy Level (1-7)

#### 1. Take a string from user and replace all spaces with underscore _ usingreplace() method.

```python
name = input("Enter String: ")

print(name.replace(" ","_"))

             
```

#### 2. Take any string and remove leading and trailing spaces using strip().

```python
name = input("Enter String: ")
print(name.strip())
```

#### 3. Take a sentence and check whether it starts with "Hello" using startswith().

```python
sent = input("Enter Sentence: ")

sent.startswith("Hello")
```

#### 4. Take a string and check if it ends with ".com" using endswith().

```python
sent = input("Enter Sentence: ")

sent.endswith(".com")
```

#### 5. Take a comma-separated string like "apple,banana,mango" and split it using split().

```python
string = ("apple,banana,mango")

string.split(",")
```

#### 6. Create a list of 5 fruits and join them using space " " with join().

```python
list = ['apple','mango','banana','orange','graphs']
(" ".join(list))
```

#### 7. Create a list: [" Python ", " is ", " awesome "]. Strip each word and join them.

```python
list =["Python", "is", "awesome"]
"" .join(list)
```

#### 11. Take a string "Hello World Python" . Replace "World" with "Everyone" and then split the result

```python
a = ("Hello World Python")
b=(a.replace("World","Everyone"))
print(b.split(" "))
```

#### 12. Create a list of 5 city names from user input. Join them with new line "\n" and print.


```python
x= input("Enter 5 city Name: ")
z=x.split(",")

y=("\n".join(z))
print(y)

```

#### 13. Take a string " hello world ". Strip it, convert to title case, and check if it starts with "Hello".

```python
name = " hello world "

name=name.strip()
print(name)
name=name.title()
print(name)
name.startswith("Hello")
```

#### 14.Take a sentence and replace the word "bad" with "good" after splitting and joining.

```python
sent = "Nayan is bad boy"
sent.split()
"".join(sent)
sent.replace("bad","good")
```

#### 15. Take a list ["10", "20", "30", "40"] and join them with "-" using join()
