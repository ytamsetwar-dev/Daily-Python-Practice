### If Else & Nested If Else 

```python
age= 19
Id = True

if age >= 18:
    if Id == True:
        print("Eligible for Vote:")
    else:
        print("Bring Id Card")
else :
    print("Not allowed")
```

```python
A=10
B=20
C=30

if A>=B:
    if A>=C:
        print("A is Greater")
    else:
        print("C is Greater")
else:
    if B <= C:
        print("C is Greater")
    else:
        print("B is Greater")
```

```python

pin = 1234
balance = 12000

upin = int(input("Enter The Pin: "))
if upin == pin:
    amount=int(input("Enter Amount: "))
    if balance >= amount:
        print("Transaction Successfully")
    else:
        print("Insufficient Balance")
else:
    print("Invalid Pin")

```
