```python
# Find prime number

for num in range(10,51):

    if num > 1:

        for i in range(2,num):

            if num % i==0:

                print(f"{num} is not Prime")
                break
            else:
                print(f"{num} is Prime")
                break
    
```

### Zip() and enumerator()


```python
# 2. The Iteration Helpers: zip() and enumerate()
   
# enumerate() attaches an automatic counter to every item in a collection. It saves you from manually creating an index = 0 variable before a loop.

# zip() acts like a physical zipper on a jacket. It pairs up items from two separate lists side-by-side.

# The Interview Trap: What if the two lists are different lengths? zip() stops early at the shortest list, completely ignoring the extra items in the longer list!
```

```python
# Ex 1: zip() 

a = [1,2,3,4]
b = ['a','b','c','d','e','f']

print(tuple(zip(a,b)))
```

```python
# Ex 2: zip()

list(enumerate([a]))
```
